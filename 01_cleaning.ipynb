```python
import pandas as pd
import numpy as np
from joblib import Memory
memory = Memory(location='./.cache', verbose=0)
import os
```

```python
data_original = pd.read_csv('vehicles.csv')

data = data_original.copy()

data = data.drop_duplicates()
```

```python
data.describe(include='all').T
```

```python
data.info()
```

```python
data.columns = data.columns.str.strip().str.lower()
```

```python
missing_pct = data.isnull().mean() * 100
missing_pct.sort_values(ascending=False, inplace=True)
missing_pct
```

##### Drop county, completely empty

```python
unique = data[data.select_dtypes(include=['object']).columns].nunique()
unique
```

##### Drop url, region_url, image_url, description, posting_date, VIN - high cardinality

```python
cols_to_drop = ['url', 'region_url', 'description', 'image_url', 'posting_date', 'vin', 'county', '']

data.drop(columns=cols_to_drop, inplace=True)
```

```python
# Save the cleaned dataset and put it in .cache 
data.to_parquet('.cache/vehicles_cleaned.parquet')
```
